## 1. Import libraries

## 2. Load the dataset

**Dataset:** Mall Customers Dataset (200 customer records, 5 columns).
This is a widely used, publicly available dataset for teaching customer
segmentation / clustering, originally distributed on Kaggle
("Mall Customer Segmentation Data"). It contains basic demographic
information (Age, Gender) together with two business-relevant behavioral
features: Annual Income and a Spending Score assigned by the mall based
on customer behavior.

**Why this dataset was chosen (feasibility analysis):**
- Small enough (200 rows) to run quickly and be fully understood by a
  beginner, yet large enough to reveal genuine cluster structure.
- No missing values, no messy formatting -> feasible for an academic
  project timeline without spending review cycles on data cleaning.
- Contains a mix of demographic (Age) and behavioral (Income, Spending
  Score) features -> good separation for K-Means and interesting noise
  points for DBSCAN.
- Numeric features are directly interpretable for business decisions
  (e.g. "high income, low spending" is an actionable segment).


```python
df = pd.read_csv("../data/raw/Mall_Customers.csv")
print("Dataset loaded.")
df.head(10)
```

## 4. Data types

`CustomerID` is a unique numeric identifier (not a real feature).
`Gender` is categorical (text).
`Age`, `Annual Income (k$)`, and `Spending Score (1-100)` are numeric
and will be the main features used for clustering.


```python
df.dtypes.to_frame(name="dtype")
```

## 5. Missing value analysis

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_report = pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
print(missing_report)
print()
print(f"Total missing values in dataset: {df.isnull().sum().sum()}")
```

**Interpretation:** the dataset has **zero missing values**, so no
imputation is required for this particular export. The preprocessing module
(`src/preprocessing.py`) still contains a missing-value-handling step so the
pipeline remains reusable on a future, messier customer export.


## 6. Duplicate record analysis

```python
n_duplicates = df.duplicated().sum()
print(f"Number of fully duplicated rows: {n_duplicates}")
n_duplicate_ids = df['CustomerID'].duplicated().sum()
print(f"Number of duplicated CustomerIDs: {n_duplicate_ids}")
```
